## 4. Crear el objeto BeautifulSoup

```python
soup = BeautifulSoup(html, "html.parser")
print(soup.title)
```

## 5. Buscar todas las secciones y los restaurantes

Cada restaurante está dentro de un `div` con la clase `restaurant`.

```python
secciones = soup.find_all("section")
print("Número de secciones:", len(secciones))

for seccion in secciones:
    print("Sección:", seccion.get("id"))
    restaurantes = seccion.find_all("div", class_="restaurant")
    print("Restaurantes en esta sección:", len(restaurantes))
    print()
```

## 6. Extraer nombre y descripción de cada restaurante

```python
for seccion in soup.find_all("section"):
    for restaurante in seccion.find_all("div", class_="restaurant"):
        nombre = restaurante.find("h3").get_text(strip=True)
        descripcion = restaurante.find("p").get_text(strip=True)
        print("Nombre:", nombre)
        print("Descripción:", descripcion)
        print()
```

## 7. Extraer imagen, enlace y guardarlos en CSV

```python
import csv

lista_restaurantes = []

for seccion in soup.find_all("section"):
    for restaurante in seccion.find_all("div", class_="restaurant"):
        nombre = restaurante.find("h3").get_text(strip=True)
        descripcion = restaurante.find("p").get_text(strip=True)
        enlace = restaurante.find("a", class_="read-more").get("href")
        imagen = restaurante.find("img").get("src")
        seccion_nombre = seccion.get("id", "sin_categoria")

        registro = {
            "seccion": seccion_nombre,
            "nombre": nombre,
            "descripcion": descripcion,
            "enlace": enlace,
            "imagen": imagen
        }

        lista_restaurantes.append(registro)

        print("Sección:", seccion_nombre)
        print("Nombre:", nombre)
        print("Descripción:", descripcion)
        print("Enlace:", enlace)
        print("Imagen:", imagen)
        print("-" * 50)

ruta_csv = r"C:\Users\marke\Desktop\CLASE\API REST\Scrapy\restaurante\restaurantes.csv"

with open(ruta_csv, "w", newline="", encoding="utf-8") as archivo:
    writer = csv.DictWriter(archivo, fieldnames=["seccion", "nombre", "descripcion", "enlace", "imagen"])
    writer.writeheader()
    writer.writerows(lista_restaurantes)

print("\nCSV guardado en:", ruta_csv)
```

## 10. Extraer todos los restaurantes

```python
from bs4 import BeautifulSoup

ruta = r"C:\Users\marke\Desktop\CLASE\API REST\Scrapy\restaurante\restaurantes.html"

with open(ruta, "r", encoding="utf-8") as f:
    html = f.read()

soup = BeautifulSoup(html, "html.parser")
restaurantes = []

for seccion in soup.find_all("section"):
    categoria = seccion.get("id", "sin_categoria")
    for restaurante in seccion.find_all("div", class_="restaurant"):
        nombre = restaurante.find("h3").get_text(strip=True)
        descripcion = restaurante.find("p").get_text(strip=True)
        enlace = restaurante.find("a", class_="read-more").get("href")
        imagen = restaurante.find("img").get("src")

        restaurantes.append({
            "categoria": categoria,
            "nombre": nombre,
            "descripcion": descripcion,
            "enlace": enlace,
            "imagen": imagen
        })

for restaurante in restaurantes:
    print(restaurante)
```
