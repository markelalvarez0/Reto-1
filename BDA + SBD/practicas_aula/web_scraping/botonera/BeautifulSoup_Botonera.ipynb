## 7. Extraer datos de jugadores y guardarlos en CSV


```python
import csv

lista_equipos = []

tabla_clasificacion = soup.select("table")[0]

for fila in tabla_clasificacion.select("tbody tr"):
    celdas = fila.find_all("td")
    if len(celdas) == 7:
        equipo = {
            "posicion": celdas[0].get_text(strip=True),
            "equipo": celdas[1].get_text(strip=True),
            "partidos_jugados": celdas[2].get_text(strip=True),
            "victorias": celdas[3].get_text(strip=True),
            "empates": celdas[4].get_text(strip=True),
            "derrotas": celdas[5].get_text(strip=True),
            "puntos": celdas[6].get_text(strip=True),
        }
        lista_equipos.append(equipo)

        print(equipo)

ruta_csv = r"C:\Users\marke\Desktop\CLASE\API REST\Scrapy\futbol\clasificacion_futbol.csv"

with open(ruta_csv, "w", newline="", encoding="utf-8") as archivo:
    writer = csv.DictWriter(archivo, fieldnames=["posicion", "equipo", "partidos_jugados", "victorias", "empates", "derrotas", "puntos"])
    writer.writeheader()
    writer.writerows(lista_equipos)


print("\nCSV guardado en:", ruta_csv)
```

## 12. Extraer datos de la botonera

Ahora vamos a trabajar con los archivos HTML de la carpeta `Scrapy/botonera` y a extraer la información de cada curso: título, descripción, temario, imagen y enlaces de navegación.

Este patrón se usa cuando tienes varias páginas con la misma estructura y quieres recopilar datos de todas a la vez.

```python
from bs4 import BeautifulSoup
import csv
import json
from pathlib import Path

base_dir = Path(r"C:\Users\marke\Desktop\CLASE\API REST\Scrapy\botonera")
archivos_html = [
    base_dir / "pagina1.html",
    base_dir / "pagina2.html",
    base_dir / "pagina3.html",
    base_dir / "pagina4.html",
    base_dir / "pagina5.html",
]

print("Archivos a procesar:", [p.name for p in archivos_html])

cursos = []

for ruta in archivos_html:
    with open(ruta, "r", encoding="utf-8") as f:
        html = f.read()

    soup = BeautifulSoup(html, "html.parser")

    titulo = soup.find("h1").get_text(strip=True)
    descripcion = soup.find("section").find("p").get_text(" ", strip=True)
    imagen = soup.find("img")["src"] if soup.find("img") else ""
    temario = [li.get_text(strip=True) for li in soup.select("ul li")]
    enlaces = [a.get("href") for a in soup.select("nav a")]

    curso = {
        "titulo": titulo,
        "descripcion": descripcion,
        "imagen": imagen,
        "temario": temario,
        "enlaces": enlaces,
        "archivo": ruta.name,
    }

    cursos.append(curso)
    print(f"Procesado: {ruta.name} -> {titulo}")

ruta_csv = base_dir / "cursos_botonera.csv"
with open(ruta_csv, "w", newline="", encoding="utf-8") as archivo:
    writer = csv.DictWriter(
        archivo,
        fieldnames=["titulo", "descripcion", "imagen", "temario", "enlaces", "archivo"]
    )
    writer.writeheader()
    for curso in cursos:
        fila = curso.copy()
        fila["temario"] = " | ".join(curso["temario"])
        fila["enlaces"] = " | ".join(curso["enlaces"])
        writer.writerow(fila)

ruta_json = base_dir / "cursos_botonera.json"
with open(ruta_json, "w", encoding="utf-8") as archivo:
    json.dump(cursos, archivo, ensure_ascii=False, indent=4)

print(f"\nSe han procesado {len(cursos)} archivos.")
print(f"CSV guardado en: {ruta_csv}")
print(f"JSON guardado en: {ruta_json}")
```
