# Ejercicios con ficheros CSV

Este notebook permite practicar la lectura, filtrado, modificación y escritura de datos CSV con Python.

Ejecuta las celdas en orden, de arriba abajo. Los ejercicios 11, 12 y 13 modifican la lista `alumnos` en memoria.

Las funciones y elementos más importantes de la librería csv son:

csv.reader(): lee cada fila como una lista.
csv.DictReader(): lee cada fila como un diccionario utilizando las cabeceras.
csv.writer(): permite escribir filas utilizando listas.
csv.DictWriter(): permite escribir filas utilizando diccionarios.
writerow(): escribe una única fila.
writerows(): escribe varias filas.
writeheader(): escribe las cabeceras definidas en fieldnames.

También son importantes los modos de apertura:

"r": leer el CSV.
"w": crear o sobrescribir el CSV.
"a": añadir filas al final.
open("alumnos.csv", "r", encoding="utf-8", newline="")

Y algunos parámetros habituales:

delimiter=";": indica el separador.
fieldnames=campos: define los nombres y el orden de las columnas.
newline="": evita líneas vacías adicionales, especialmente en Windows.
encoding="utf-8": permite trabajar correctamente con tildes y caracteres especiales.

## 1. Importar la librería CSV


```python
import csv
```

## 2. Crear el fichero alumnos.csv

Esta celda genera automáticamente el fichero de ejemplo en la misma carpeta que el notebook.


```python
contenido_csv = """id,nombre,edad,curso,nota,activo
1,Ane,20,DAM,8.5,True
2,Jon,22,DAW,6.7,True
3,Maialen,19,DAM,4.2,False
4,Unai,21,ASIR,7.8,True
"""

with open("alumnos.csv", "w", encoding="utf-8", newline="") as fichero:
    fichero.write(contenido_csv)

print("Se ha creado el fichero alumnos.csv")
```

## 3. Cargar el fichero CSV

`DictReader` utiliza la cabecera del CSV como claves de cada diccionario. Todos los valores se leen inicialmente como texto, por lo que convertimos los campos numéricos y booleanos.


```python
alumnos = []

with open("alumnos.csv", "r", encoding="utf-8", newline="") as fichero:
    lector = csv.DictReader(fichero)

    for fila in lector:
        fila["id"] = int(fila["id"])
        fila["edad"] = int(fila["edad"])
        fila["nota"] = float(fila["nota"])
        fila["activo"] = fila["activo"] == "True"
        alumnos.append(fila)

print("Número de alumnos:", len(alumnos))
```

## Ejercicio 1. Mostrar todo el contenido


```python
for alumno in alumnos:
    print(alumno)
```

## Ejercicio 2. Mostrar solo los nombres


```python
for alumno in alumnos:
    print(alumno["nombre"])
```

## Ejercicio 3. Mostrar el nombre y la nota


```python
for alumno in alumnos:
    print(alumno["nombre"], "-", alumno["nota"])
```

## Ejercicio 4. Mostrar el alumnado aprobado


```python
for alumno in alumnos:
    if alumno["nota"] >= 5:
        print(alumno["nombre"])
```

## Ejercicio 5. Mostrar el alumnado de DAM


```python
for alumno in alumnos:
    if alumno["curso"] == "DAM":
        print(alumno["nombre"])
```

## Ejercicio 6. Mostrar el alumnado activo


```python
for alumno in alumnos:
    if alumno["activo"]:
        print(alumno["nombre"])
```

## Ejercicio 7. Calcular la nota media


```python
suma = 0

for alumno in alumnos:
    suma += alumno["nota"]

media = suma / len(alumnos)
print("Nota media:", media)
```

## Ejercicio 8. Buscar un alumno por nombre


```python
nombre_buscar = "Jon"

for alumno in alumnos:
    if alumno["nombre"] == nombre_buscar:
        print(alumno)
```

## Ejercicio 9. Buscar al alumno con mayor nota


```python
mejor = alumnos[0]

for alumno in alumnos:
    if alumno["nota"] > mejor["nota"]:
        mejor = alumno

print(mejor["nombre"], mejor["nota"])
```

## Ejercicio 10. Contar alumnos por curso


```python
cursos = {}

for alumno in alumnos:
    curso = alumno["curso"]

    if curso in cursos:
        cursos[curso] += 1
    else:
        cursos[curso] = 1

print(cursos)
```

## Ejercicio 11. Modificar la nota de un alumno

La modificación se realiza inicialmente sobre la lista `alumnos`. Después se reescribe el CSV completo para conservar el cambio.


```python
for alumno in alumnos:
    if alumno["nombre"] == "Maialen":
        alumno["nota"] = 5.5

campos = ["id", "nombre", "edad", "curso", "nota", "activo"]

with open("alumnos.csv", "w", encoding="utf-8", newline="") as fichero:
    escritor = csv.DictWriter(fichero, fieldnames=campos)
    escritor.writeheader()
    escritor.writerows(alumnos)

print(alumnos)
```

## Ejercicio 12. Añadir un alumno nuevo

Se añade el nuevo diccionario a la lista y también como una nueva fila de `alumnos.csv`.


```python
nuevo_alumno = {
    "id": 5,
    "nombre": "Nerea",
    "edad": 23,
    "curso": "DAW",
    "nota": 9.1,
    "activo": True
}

alumnos.append(nuevo_alumno)

with open("alumnos.csv", "a", encoding="utf-8", newline="") as fichero:
    campos = ["id", "nombre", "edad", "curso", "nota", "activo"]
    escritor = csv.DictWriter(fichero, fieldnames=campos)
    escritor.writerow(nuevo_alumno)

print(alumnos)
```

## Ejercicio 13. Eliminar un alumno

Para eliminar un registro del CSV se crea una lista sin Jon y se reescribe el fichero completo.


```python
alumnos_filtrados = []

for alumno in alumnos:
    if alumno["nombre"] != "Jon":
        alumnos_filtrados.append(alumno)

alumnos = alumnos_filtrados
campos = ["id", "nombre", "edad", "curso", "nota", "activo"]

with open("alumnos.csv", "w", encoding="utf-8", newline="") as fichero:
    escritor = csv.DictWriter(fichero, fieldnames=campos)
    escritor.writeheader()
    escritor.writerows(alumnos)

print(alumnos)
```

## Ejercicio 14. Crear un CSV solamente con los aprobados

Los datos filtrados se guardan en `aprobados.csv`.


```python
aprobados = []

for alumno in alumnos:
    if alumno["nota"] >= 5:
        aprobados.append(alumno)

campos = ["id", "nombre", "edad", "curso", "nota", "activo"]

with open("aprobados.csv", "w", encoding="utf-8", newline="") as fichero:
    escritor = csv.DictWriter(fichero, fieldnames=campos)
    escritor.writeheader()
    escritor.writerows(aprobados)

print("Se ha creado el fichero aprobados.csv")
```

## Ejercicio 15. Leer un CSV utilizando csv.reader

`csv.reader` devuelve cada fila como una lista. `DictReader` devuelve un diccionario y resulta más cómodo cuando el CSV tiene cabeceras.


```python
with open("alumnos.csv", "r", encoding="utf-8", newline="") as fichero:
    lector = csv.reader(fichero)

    for fila in lector:
        print(fila)
```
