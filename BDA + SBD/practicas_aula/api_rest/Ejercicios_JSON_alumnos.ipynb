# Ejercicios con ficheros JSON

En este notebook se trabaja con una lista de alumnos almacenada en un fichero JSON. Ejecuta las celdas en orden, de arriba abajo.

> **Importante:** los ejercicios 11, 12 y 13 modifican la lista `alumnos`. Para recuperar los datos iniciales, vuelve a ejecutar las celdas de creación y carga del fichero.


## 1. Crear el fichero de ejemplo

Esta celda crea `alumnos.json` en la misma carpeta que el notebook. En JSON los booleanos se escriben como `true` y `false`, pero en Python se utilizan `True` y `False`.


```python
import json

datos_iniciales = [
    {"id": 1, "nombre": "Ane", "edad": 20, "curso": "DAM", "nota": 8.5, "activo": True},
    {"id": 2, "nombre": "Jon", "edad": 22, "curso": "DAW", "nota": 6.7, "activo": True},
    {"id": 3, "nombre": "Maialen", "edad": 19, "curso": "DAM", "nota": 4.2, "activo": False},
    {"id": 4, "nombre": "Unai", "edad": 21, "curso": "ASIR", "nota": 7.8, "activo": True}
]

with open("alumnos.json", "w", encoding="utf-8") as fichero:
    json.dump(datos_iniciales, fichero, indent=4, ensure_ascii=False)

print("Se ha creado el fichero alumnos.json")
```

## 2. Cargar el fichero JSON

`json.load()` convierte el contenido del fichero en una estructura de datos de Python. En este caso devuelve una lista de diccionarios.


```python
with open("alumnos.json", "r", encoding="utf-8") as fichero:
    alumnos = json.load(fichero)

print("Número de alumnos:", len(alumnos))
```

## Ejercicio 1. Mostrar todo el contenido


```python
print(alumnos)
```

## Ejercicio 2. Mostrar solo los nombres


```python
for alumno in alumnos:
    print(alumno["nombre"])
```

## Ejercicio 3. Mostrar el nombre y la nota


```python
for alumno in alumnos:
    print(alumno["nombre"], "-", alumno["nota"])
```

## Ejercicio 4. Mostrar el alumnado aprobado


```python
for alumno in alumnos:
    if alumno["nota"] >= 5:
        print(alumno["nombre"])
```

## Ejercicio 5. Mostrar el alumnado de DAM


```python
for alumno in alumnos:
    if alumno["curso"] == "DAM":
        print(alumno["nombre"])
```

## Ejercicio 6. Mostrar el alumnado activo


```python
for alumno in alumnos:
    if alumno["activo"]:
        print(alumno["nombre"])
```

## Ejercicio 7. Calcular la nota media


```python
suma = 0

for alumno in alumnos:
    suma += alumno["nota"]

media = suma / len(alumnos)
print("Nota media:", media)
```

## Ejercicio 8. Buscar un alumno por nombre


```python
nombre_buscar = "Jon"

for alumno in alumnos:
    if alumno["nombre"] == nombre_buscar:
        print(alumno)
```

## Ejercicio 9. Buscar al alumno con mayor nota


```python
mejor = alumnos[0]

for alumno in alumnos:
    if alumno["nota"] > mejor["nota"]:
        mejor = alumno

print(mejor["nombre"], mejor["nota"])
```

## Ejercicio 10. Contar alumnos por curso


```python
cursos = {}

for alumno in alumnos:
    curso = alumno["curso"]

    if curso in cursos:
        cursos[curso] += 1
    else:
        cursos[curso] = 1

print(cursos)
```

## Ejercicio 11. Modificar la nota de un alumno

Después de modificar la lista, los cambios se guardan en `alumnos.json`.


```python
for alumno in alumnos:
    if alumno["nombre"] == "Maialen":
        alumno["nota"] = 5.5

with open("alumnos.json", "w", encoding="utf-8") as fichero:
    json.dump(alumnos, fichero, indent=4, ensure_ascii=False)

print(alumnos)
```

## Ejercicio 12. Añadir un alumno nuevo

Se añade un diccionario a la lista mediante `append()` y después se actualiza el fichero.


```python
nuevo_alumno = {
    "id": 5,
    "nombre": "Nerea",
    "edad": 23,
    "curso": "DAW",
    "nota": 9.1,
    "activo": True
}

alumnos.append(nuevo_alumno)

with open("alumnos.json", "w", encoding="utf-8") as fichero:
    json.dump(alumnos, fichero, indent=4, ensure_ascii=False)

print(alumnos)
```

## Ejercicio 13. Eliminar un alumno

El `break` detiene la búsqueda cuando se encuentra y elimina a Jon.


```python
for alumno in alumnos:
    if alumno["nombre"] == "Jon":
        alumnos.remove(alumno)
        break

with open("alumnos.json", "w", encoding="utf-8") as fichero:
    json.dump(alumnos, fichero, indent=4, ensure_ascii=False)

print(alumnos)
```

## Ejercicio 14. Crear un nuevo JSON con los aprobados

Se recorren los alumnos actuales y se guardan en otro fichero únicamente quienes tienen una nota igual o superior a 5.


```python
aprobados = []

for alumno in alumnos:
    if alumno["nota"] >= 5:
        aprobados.append(alumno)

with open("aprobados.json", "w", encoding="utf-8") as fichero:
    json.dump(aprobados, fichero, indent=4, ensure_ascii=False)

print("Se ha creado el fichero aprobados.json")
```

## Comprobar el contenido de aprobados.json


```python
with open("aprobados.json", "r", encoding="utf-8") as fichero:
    aprobados_guardados = json.load(fichero)

for alumno in aprobados_guardados:
    print(alumno["nombre"], alumno["nota"])
```
