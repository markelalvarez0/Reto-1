### Ejercicio 4
Obtén todos los posts del usuario con `userId = 2` utilizando `params=`.

````python
BASE_POSTS = "https://jsonplaceholder.typicode.com/posts"

respuesta = requests.get(BASE_POSTS, params={"userId": 2})
posts = respuesta.json()

print(posts)
````

### Ejercicio 5
Muestra el id y el título de los posts obtenidos.

````python
for post in posts:
    print(post["id"], "-", post["title"])
````

### Ejercicio 6
Cuenta cuántos posts tiene cada usuario.

````python
respuesta = requests.get(BASE_POSTS)
posts = respuesta.json()

contadores = {}

for post in posts:
    user_id = post["userId"]
    
    if user_id in contadores:
        contadores[user_id] += 1
    else:
        contadores[user_id] = 1

for user_id, cantidad in contadores.items():
    print("Usuario", user_id, ":", cantidad, "posts")
````

### Ejercicio 7
Obtén los comentarios del post con `id = 5` y muestra nombre y email.

## Parte 2. Open-Meteo

API base: `https://api.open-meteo.com/v1/forecast`

````python
OPEN_METEO = "https://api.open-meteo.com/v1/forecast"
LAT = 43.263
LON = -2.935
````

### Ejercicio 9
Consulta la temperatura actual de Bilbao.

````python
parametros = {
    "latitude": LAT,
    "longitude": LON,
    "current": "temperature_2m"
}

respuesta = requests.get(OPEN_METEO, params=parametros)

respuesta.raise_for_status()

datos = respuesta.json()

print("Temperatura actual:", datos["current"]["temperature_2m"], "C")
````

### Ejercicio 10
Obtén temperatura, humedad relativa y velocidad del viento.

````python
parametros = {
    "latitude": LAT,
    "longitude": LON,
    "current": "temperature_2m,relative_humidity_2m,wind_speed_10m"
}

respuesta = requests.get(OPEN_METEO, params=parametros)

respuesta.raise_for_status()

datos = respuesta.json()

print("- Temperatura actual:", datos["current"]["temperature_2m"], "C", '- Humedad Relativa:', datos['current']['relative_humidity_2m'], '- Velocidad del viento:', datos['current']['wind_speed_10m'])
````

### Ejercicio 11
Obtén la predicción horaria de temperatura y muestra las primeras 10 horas.

````python
parametros = {
    "latitude": LAT,
    "longitude": LON,
    "hourly": "temperature_2m",
    "timezone": "Europe/Madrid"
}

datos = requests.get(OPEN_METEO, params=parametros).json()

for i in range(10):
    print(
        datos["hourly"]["time"][i], '-',
        datos["hourly"]["temperature_2m"][i]
    )
````

### Ejercicio 12
Obtén la temperatura máxima y mínima diaria.

````python
parametros = {
    "latitude": LAT,
    "longitude": LON,
    "daily": "temperature_2m_max,temperature_2m_min",
    "timezone": "Europe/Madrid"
}

datos = requests.get(OPEN_METEO, params=parametros).json()

for i in range(len(datos["daily"]["time"])):
    print(
        datos["daily"]["time"][i], '-',
        datos["daily"]["temperature_2m_max"][i], '-',
        datos["daily"]["temperature_2m_min"][i]
    )
````
