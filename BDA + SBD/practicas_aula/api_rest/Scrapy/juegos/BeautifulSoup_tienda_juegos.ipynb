## 5. Buscar todas las secciones y los juegos

Cada juego está dentro de un `div` con la clase `juego`.

```python
secciones = soup.find_all("section")
print("Número de secciones:", len(secciones))

for seccion in secciones:
    print("Sección:", seccion.get("id"))
    juegos = seccion.find_all("div", class_="juego")
    print("Juegos en esta sección:", len(juegos))
    print()
```

## 6. Extraer nombre y precio de cada juego

```python
for seccion in soup.find_all("section"):
    for juego in seccion.find_all("div", class_="juego"):
        nombre = juego.find("h3").get_text(strip=True)
        precio = juego.find("p", class_="precio").get_text(strip=True)
        print("Nombre:", nombre)
        print("Precio:", precio)
        print()
```

## 7. Extraer id, imagen y enlace de compra y guardarlos en CSV

```python
import csv

lista_juegos = []

for seccion in soup.find_all("section"):
    for juego in seccion.find_all("div", class_="juego"):
        nombre = juego.find("h3").get_text(strip=True)
        precio = juego.find("p", class_="precio").get_text(strip=True)
        id_juego = juego.get("data-id")
        enlace = juego.find("a", class_="comprar-btn").get("href")
        imagen = juego.find("img").get("src")

        registro = {
            "seccion": seccion.get("id", "sin_categoria"),
            "nombre": nombre,
            "precio": precio,
            "enlace": enlace,
            "imagen": imagen
        }

        lista_juegos.append(registro)

        print("ID:", id_juego)
        print("Nombre:", nombre)
        print("Precio:", precio)
        print("Enlace:", enlace)
        print("Imagen:", imagen)
        print("-" * 50)

ruta_csv = r"C:\Users\marke\Desktop\CLASE\API REST\Scrapy\juegos\juegos_tienda.csv"

with open(ruta_csv, "w", newline="", encoding="utf-8") as archivo:
    writer = csv.DictWriter(archivo, fieldnames=["seccion", "nombre", "precio", "enlace", "imagen"])
    writer.writeheader()
    writer.writerows(lista_juegos)

print("\nCSV guardado en:", ruta_csv)
```

## 8. Leer el CSV y convertirlo a JSON

```python
import csv
import json

ruta_csv = r"C:\Users\marke\Desktop\CLASE\API REST\Scrapy\juegos\juegos_tienda.csv"

with open(ruta_csv, "r", encoding="utf-8", newline="") as archivo:
    lector = csv.DictReader(archivo)
    juegos = list(lector)

for juego in juegos:
    print(juego)

ruta_json = r"C:\Users\marke\Desktop\CLASE\API REST\Scrapy\juegos_tienda.json"

with open(ruta_json, "w", encoding="utf-8") as archivo:
    json.dump(juegos, archivo, ensure_ascii=False, indent=4)

print("\nJSON guardado en:", ruta_json)
```

## 10. Extraer todos los juegos de la tienda

Esta es la solución enfocada a la tienda, que recoge cada juego con su categoría, id, nombre, precio, enlace e imagen.

```python
from bs4 import BeautifulSoup

ruta = r"C:\Users\marke\Desktop\CLASE\API REST\Scrapy\juegos\tienda_juegos (1).html"

with open(ruta, "r", encoding="utf-8") as f:
    html = f.read()

soup = BeautifulSoup(html, "html.parser")

juegos = []

for seccion in soup.find_all("section"):
    categoria = seccion.get("id", "sin_categoria")

    for juego in seccion.select(".juego"):
        nombre = juego.find("h3").get_text(strip=True)
        precio = juego.select_one(".precio").get_text(strip=True)
        enlace = juego.select_one("a.comprar-btn").get("href")
        imagen = juego.select_one("img").get("src")
        id_juego = juego.get("data-id")

        juegos.append({
            "categoria": categoria,
            "id": id_juego,
            "nombre": nombre,
            "precio": precio,
            "enlace": enlace,
            "imagen": imagen
        })

# Mostrar resultado
for juego in juegos:
    print(juego)
```

## Resumen de funciones importantes

| Función o elemento | Utilidad |
|---|---|
| `BeautifulSoup(contenido, "html.parser")` | Convierte el texto HTML en una estructura navegable. |
| `find()` | Devuelve la primera etiqueta encontrada. |
| `find_all()` | Devuelve todas las etiquetas encontradas. |
| `class_="curso"` | Filtra por una clase CSS. |
| `etiqueta["title"]` | Obtiene el valor de un atributo. |
| `get_text(strip=True)` | Obtiene el texto y elimina espacios sobrantes. |
| `prettify()` | Muestra el HTML con una estructura legible. |
