## 2. Importar BeautifulSoup


```python
from bs4 import BeautifulSoup
```

## 3. Crear el fichero ejemplo.html

Esta celda crea un HTML sencillo con una categoría y tres cursos.


```python
contenido_html = """<!DOCTYPE html>
<html lang="es">
<head>
    <meta charset="UTF-8">
    <title>Oferta formativa</title>
</head>
<body>
    <div class="categoria" title="Cursos de informática">
        <h3>Informática y comunicaciones</h3>

        <div class="curso" title="Desarrollo de Aplicaciones Multiplataforma">
            DAM
        </div>

        <div class="curso" title="Desarrollo de Aplicaciones Web">
            DAW
        </div>

        <div class="curso" title="Administración de Sistemas Informáticos en Red">
            ASIR
        </div>
    </div>
</body>
</html>
"""

with open("ejemplo.html", "w", encoding="utf-8") as file:
    file.write(contenido_html)

print("Se ha creado el fichero ejemplo.html")
```

## 4. Leer el archivo HTML

`read()` recupera todo el contenido del fichero como una cadena de texto.


```python
with open("ejemplo.html", "r", encoding="utf-8") as file:
    contenido = file.read()

print(contenido)
```

## 5. Crear el objeto BeautifulSoup

`html.parser` indica que el contenido debe interpretarse como HTML.


```python
soup = BeautifulSoup(contenido, "html.parser")
print(soup.prettify())
```

## 6. Extraer un div por su clase

`find()` devuelve la primera etiqueta que cumple la condición.


```python
div_categoria = soup.find("div", class_="categoria")
print(div_categoria)
```

## 7. Obtener un atributo del div

Los atributos de una etiqueta pueden consultarse utilizando corchetes, de forma parecida a un diccionario.


```python
titulo_categoria = div_categoria["title"]
print(titulo_categoria)
```

## 8. Extraer todos los div con la clase curso

`find_all()` devuelve una lista con todas las etiquetas encontradas.


```python
divs_curso = soup.find_all("div", class_="curso")

for curso in divs_curso:
    print(curso)
```

## 9. Obtener el texto y el atributo title de cada curso

`get_text(strip=True)` devuelve el texto interior y elimina los espacios y saltos de línea sobrantes.


```python
for curso in divs_curso:
    curso_titulo = curso.get_text(strip=True)
    curso_title_attr = curso["title"]

    print("Código:", curso_titulo)
    print("Nombre completo:", curso_title_attr)
    print()
```

## 10. Obtener el texto del h3

La búsqueda se realiza dentro de `div_categoria`, no en todo el documento.


```python
h3_titulo = div_categoria.find("h3").get_text(strip=True)
print(h3_titulo)
```

## Resumen de funciones importantes


| Función o elemento | Utilidad |
|---|---|
| `BeautifulSoup(contenido, "html.parser")` | Convierte el texto HTML en una estructura navegable. |
| `find()` | Devuelve la primera etiqueta encontrada. |
| `find_all()` | Devuelve todas las etiquetas encontradas. |
| `class_="curso"` | Filtra por una clase CSS. |
| `etiqueta["title"]` | Obtiene el valor de un atributo. |
| `get_text(strip=True)` | Obtiene el texto y elimina espacios sobrantes. |
| `prettify()` | Muestra el HTML con una estructura legible. |
