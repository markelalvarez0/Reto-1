## 4. Crear el objeto BeautifulSoup

## 5. Buscar todas las tablas y ver sus datos

Cada tabla contiene una sección distinta: clasificación y estadísticas.


```python
tablas = soup.find_all("table")
print("Número de tablas:", len(tablas))

for i, tabla in enumerate(tablas, start=1):
    titulo = tabla.find("caption").get_text(strip=True)
    filas = tabla.find_all("tr")
    print(f"Tabla {i}: {titulo}")
    print("Filas totales:", len(filas))
    print()
```

## 6. Extraer la clasificación de equipos


```python
tabla_clasificacion = soup.find("table")
headers = [th.get_text(strip=True) for th in tabla_clasificacion.find_all("th")]
print("Cabeceras:", headers)

for fila in tabla_clasificacion.select("tbody tr"):
    valores = [td.get_text(strip=True) for td in fila.find_all("td")]
    print(valores)
```

## 7. Extraer datos de jugadores y guardarlos en CSV


```python
import csv

lista_equipos = []

tabla_clasificacion = soup.select("table")[0]

for fila in tabla_clasificacion.select("tbody tr"):
    celdas = fila.find_all("td")
    if len(celdas) == 7:
        equipo = {
            "posicion": celdas[0].get_text(strip=True),
            "equipo": celdas[1].get_text(strip=True),
            "partidos_jugados": celdas[2].get_text(strip=True),
            "victorias": celdas[3].get_text(strip=True),
            "empates": celdas[4].get_text(strip=True),
            "derrotas": celdas[5].get_text(strip=True),
            "puntos": celdas[6].get_text(strip=True),
        }
        lista_equipos.append(equipo)

        print(equipo)

ruta_csv = r"C:\Users\marke\Desktop\CLASE\API REST\Scrapy\futbol\clasificacion_futbol.csv"

with open(ruta_csv, "w", newline="", encoding="utf-8") as archivo:
    writer = csv.DictWriter(archivo, fieldnames=["posicion", "equipo", "partidos_jugados", "victorias", "empates", "derrotas", "puntos"])
    writer.writeheader()
    writer.writerows(lista_equipos)


print("\nCSV guardado en:", ruta_csv)
```

## 8. Leer el CSV y convertirlo a JSON


```python
import csv
import json

ruta_csv = r"C:\Users\marke\Desktop\CLASE\API REST\Scrapy\futbol\clasificacion_futbol.csv"

with open(ruta_csv, "r", encoding="utf-8", newline="") as archivo:
    lector = csv.DictReader(archivo)
    equipos = list(lector)

for equipo in equipos:
    print(equipo)

ruta_json = r"C:\Users\marke\Desktop\CLASE\API REST\Scrapy\futbol\clasificacion_futbol.json"

with open(ruta_json, "w", encoding="utf-8") as archivo:
    json.dump(equipos, archivo, ensure_ascii=False, indent=4)

print("\nJSON guardado en:", ruta_json)
```

## 10. Extraer todos los equipos de la clasificación

Esta es la solución enfocada a la tabla de fútbol, que recoge cada equipo con su posición, nombre y estadísticas.


```python
from bs4 import BeautifulSoup

ruta = r"C:\Users\marke\Desktop\CLASE\API REST\Scrapy\futbol\tabla_futbol.html"

with open(ruta, "r", encoding="utf-8") as f:
    html = f.read()

soup = BeautifulSoup(html, "html.parser")

equipos = []

tabla_clasificacion = soup.select("table")[0]

for fila in tabla_clasificacion.select("tbody tr"):
    celdas = fila.find_all("td")
    if len(celdas) == 7:
        equipos.append({
            "posicion": celdas[0].get_text(strip=True),
            "equipo": celdas[1].get_text(strip=True),
            "partidos_jugados": celdas[2].get_text(strip=True),
            "victorias": celdas[3].get_text(strip=True),
            "empates": celdas[4].get_text(strip=True),
            "derrotas": celdas[5].get_text(strip=True),
            "puntos": celdas[6].get_text(strip=True),
        })

for equipo in equipos:

    print(equipo)
```

## Resumen de funciones importantes

| Función o elemento | Utilidad |
|---|---|
| `BeautifulSoup(contenido, "html.parser")` | Convierte el texto HTML en una estructura navegable. |
| `find()` | Devuelve la primera etiqueta encontrada. |
| `find_all()` | Devuelve todas las etiquetas encontradas. |
| `class_="curso"` | Filtra por una clase CSS. |
| `etiqueta["title"]` | Obtiene el valor de un atributo. |
| `get_text(strip=True)` | Obtiene el texto y elimina espacios sobrantes. |
| `prettify()` | Muestra el HTML con una estructura legible. |
