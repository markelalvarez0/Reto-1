## PRODUCTOR

```python
from confluent_kafka import Producer
import socket

conf = {'bootstrap.servers':  "10.22.87.215:9092",
        'client.id': socket.gethostname()}

producer = Producer(conf)
topic="prueba"
```

```python
def confirmar(err, msg):
    if err:
        print("Error al enviar:", err)
    else:
        print("Enviado:", msg.value().decode("utf-8"))

for i in range(10):
    producer.produce(
        topic,
        key=f"key{i}",
        value=f"Value: {20+i}",
        callback=confirmar
    )

producer.flush(15)
```

## CONSUMIDOR

```python
from confluent_kafka import Consumer

conf = {
    'bootstrap.servers': '10.22.87.215:9092',
    'group.id': 'grupo_prueba_2026_01',
    'auto.offset.reset': 'earliest'
}

consumer = Consumer(conf)
consumer.subscribe(['prueba'])

```

```python
try:
    for i in range(15):
        msg = consumer.poll(timeout=2.0)

        if msg is None:
            print("No hay mensajes")
            continue

        if msg.error():
            print("Error:", msg.error())
            continue

        print("Recibido:", msg.value().decode("utf-8"))

finally:
    consumer.close()
```

## HDFS

```python
from hdfs import InsecureClient

client = InsecureClient(
    "http://10.22.87.215:50070",
    user="VUESTRO_USUARIO"
)

#crear carpeta
client.makedirs('/user/VUESTRO_USUARIO/pruebaHDFS')
```

# Escribir en HDFS

```python
#escribir en HDFS
import json

datos = [
    {"key": "S01", "value": 25},
    {"key": "S02", "value": 28},
    {"key": "S03", "value": 22}
]

with client.write(
    '/user/VUESTRO_USUARIO/pruebaHDFS/datos.jsonl',
    encoding='utf-8',
    overwrite=True
) as writer:
    for dato in datos:
        writer.write(json.dumps(dato) + '\n')
```

# Leer del HDFS

```python
#leer en HDFS
with client.read(
    '/user/VUESTRO_USUARIO/pruebaHDFS/datos.jsonl',
    encoding='utf-8'
) as reader:
    for linea in reader:
        dato = json.loads(linea)
        print(dato)
```
